## Objective : CSV Data Import to a Database

The objective of this project is to Given a dataset containing information about students' test scores, fetch the data from an API, calculate the average score, and create a bar chart to visualize the data.

### Importing Required Libraries

```python
import requests
import pandas as pd
import matplotlib.pyplot as plt
```

### Fetching data from an external REST API

```python
url = "https://api.slingacademy.com/v1/sample-data/files/student-scores.json"
```

```python
response = requests.get(url)
```

```python
response
```

### Converting json response in python data

```python
data=response.json()
```

### A Quick Look at the Student Data

```python
data[:3]
```

### Creating a DataFrame

```python
df = pd.DataFrame(data)
df.head()
```

### Checking shape of data

```python
df.shape
```

### Looking at the Columns Names

```python
df.columns
```

### Extracting required Columns using fancing indexing

```python
students=df[['first_name', 'last_name','math_score', 'history_score', 'physics_score','chemistry_score', 'biology_score', 'english_score','geography_score']]
students.head()
```

### Checking for missing values

```python
students.isnull().sum()
```

### getting information about my columns

```python
students.info()
```

### Summary satistics value of data

```python
students.describe()
```

```python
subjects=['math_score', 'history_score', 'physics_score','chemistry_score', 'biology_score', 'english_score','geography_score']
```

```python
students[subjects]
```

### Calculating the Average Marks of each Student

```python
students["average_score"] = students[subjects].mean(axis=1).round(2)

students.head()
```

### Subject-wise Average Score

```python
subject_average = students[subjects].mean().round(2)

print(subject_average)
```

### creating a bar chart to visualize the data.


```python
subject_names = ["Math", "History", "Physics", "Chemistry","Biology", "English", "Geography"]

plt.figure(figsize=(10, 5))

plt.bar(subject_names, subject_average, color=['red','blue','green','pink','yellow','brown','orange'],edgecolor="black")

plt.title("Average marks scored by students in each Subject")
plt.xlabel("Subjects")
plt.ylabel("Subject Average Score")
plt.ylim(0, 100)

plt.tight_layout()
plt.show()
```

### Storing data in csv file

```python
df.to_csv("students.csv", index=False)
```
