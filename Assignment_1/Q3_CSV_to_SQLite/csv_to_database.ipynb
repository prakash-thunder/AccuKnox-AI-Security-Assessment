## Objective : CSV Data Import to a Database

The objective of this project is to write a Python script that reads data from a CSV file containing user information (e.g., name, email) and inserts it into a SQLite database.

### Importing Required Libraries

```python
import pandas as pd
import sqlite3
```

### loading dataset from csv file

```python
df=pd.read_csv('users.csv')
df.head()
```

### Checkinfo for missing values

```python
df.isnull().sum()
```

### Droping those missing values

```python
df = df.dropna()
```

### Checking duplicate

```python
df["email"].duplicated().sum()
```

### Droping duplicates rows

```python
df = df.drop_duplicates(subset=["email"])
```

### Previews of data after all preprocessing

```python
df.sample(10)
```

### Checking Shapes of data

```python
df.shape
```

## Establish SQLite Database Connection

```python
conn = sqlite3.connect("users.db")
cursor = conn.cursor()

print("Database connected successfully")
```

### creating table

```python
cursor.execute('''
    CREATE TABLE IF NOT EXISTS users (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT NOT NULL,
        email TEXT NOT NULL UNIQUE)''')

print("Table created successfully")
```

### Storing records

```python
user_records = []

for index, row in df.iterrows():
    name = row["name"]
    email = row["email"]

    user_records.append((name, email))
```

### inserting data in users table

```python
cursor.executemany('''
    INSERT INTO users (name, email)
    VALUES (?, ?)''', user_records)

conn.commit()
```

### Retrieving data

```python
cursor.execute("SELECT * FROM users")

records = cursor.fetchall()

records[:5]
```
